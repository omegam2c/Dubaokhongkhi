```python
import pandas as pd
import numpy as np

INPUT_FILE = "test_data_caugiay_full.csv"   
OUTPUT_FILE = "test_data_caugiay_full_attributes.csv" 

# Đọc file dữ liệu thủ công
print(f"Đang đọc file: {INPUT_FILE} ...")
df = pd.read_csv(INPUT_FILE)

# 2. Xử lý thời gian
time_col = 'time' if 'time' in df.columns else 'Timestamp'
df['Timestamp'] = pd.to_datetime(df[time_col])

df['Hour'] = df['Timestamp'].dt.hour
df['DayOfWeek'] = df['Timestamp'].dt.dayofweek
df['Month'] = df['Timestamp'].dt.month

# 3. Tính toán PM1 và UM003
if 'PM2.5' not in df.columns and 'pm2_5' in df.columns:
    df['PM2.5'] = df['pm2_5']
    
df['PM1'] = df['PM2.5'] * 0.7
df['UM003'] = df['PM2.5'] * 39.7

# 4. Tra cứu thông tin tĩnh của quận (Đang để mặc định Cầu Giấy)
DISTRICT_INFO = {
    "Cau_Giay": {"lat": 21.036, "lon": 105.795, "area": 2, "pop": 23000},
    "Dong_Da": {"lat": 21.012, "lon": 105.821, "area": 2, "pop": 37000},
    "Soc_Son": {"lat": 21.258, "lon": 105.816, "area": 1, "pop": 1100},
    "Thanh_Xuan": {"lat": 20.994, "lon": 105.799, "area": 2, "pop": 31000}
}

if 'District_ID' not in df.columns:
    df['District_ID'] = "Cau_Giay"

df['Latitude'] = df['District_ID'].apply(lambda x: DISTRICT_INFO.get(x, DISTRICT_INFO["Cau_Giay"])['lat'])
df['Longitude'] = df['District_ID'].apply(lambda x: DISTRICT_INFO.get(x, DISTRICT_INFO["Cau_Giay"])['lon'])
df['Area_Type'] = df['District_ID'].apply(lambda x: DISTRICT_INFO.get(x, DISTRICT_INFO["Cau_Giay"])['area'])
df['Pop_Density'] = df['District_ID'].apply(lambda x: DISTRICT_INFO.get(x, DISTRICT_INFO["Cau_Giay"])['pop'])

# Chuẩn hóa tên cột thời tiết
if 'Temperature' not in df.columns and 'temperature_2m' in df.columns: df['Temperature'] = df['temperature_2m']
if 'Relative_Humidity' not in df.columns and 'relative_humidity_2m' in df.columns: df['Relative_Humidity'] = df['relative_humidity_2m']
if 'Wind_Speed' not in df.columns and 'wind_speed_10m' in df.columns: df['Wind_Speed'] = df['wind_speed_10m']

# 5. Lọc đúng 15 cột
COLUMNS_ORDER = [
    'Timestamp', 'District_ID', 'Latitude', 'Longitude', 'Area_Type', 
    'Pop_Density', 'Temperature', 'Relative_Humidity', 'Wind_Speed', 
    'PM1', 'PM2.5', 'UM003', 'Hour', 'DayOfWeek', 'Month'
]
final_df = df[COLUMNS_ORDER]

# 6. Lưu file và hiển thị kết quả
final_df.to_csv(OUTPUT_FILE, index=False)
print(f"\n✅ Đã tạo thành công file: {OUTPUT_FILE}")

# Hiển thị bảng dữ liệu ra màn hình Notebook để kiểm tra
final_df.head()
```
